### Implementing K means clustering without using built in functions

```python
import numpy as np

class Kmeans:
    def __init__(self, nclusters , iters=100):
        self.nclusters = nclusters
        self.iters = iters
    
    def fit(self,x):
        self.centroids = self.centroids = x[np.random.choice(x.shape[0], self.nclusters, replace=False)]

        for _ in range(self.iters):
            assign = self.assignment(x)
            new = self.update(x,assign)
            
            if np.all(self.centroids==new):
                break

        self.centroids == new

    def assignment(self,x):
        assign = []
        for point in x:
            dists = []
            for center in self.centroids:
                dist = (np.linalg.norm(point - center)**2)
                dists.append(dist)

            index = np.argmin(dists)
            assign.append(index)
        return np.array(assign)
    
    def update(self,x,assign):
        new = []
        for i in range(self.nclusters):
            clust = x[assign==i]
            clustmean = clust.mean(axis = 0)
            new.append(clustmean)

        return np.array(new)
        
```

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer()
df_cancer = pd.DataFrame(data['data'])
df_cancer.columns = ['f' + str(i) for i in range(len(df_cancer.columns))]

km = Kmeans(nclusters=3)
km.fit(df_cancer.values) 

print(km.centroids)
print(km.centroids.shape)
```
